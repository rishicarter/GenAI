```python
def binary_search(array, num):
    """
    Time Complexity: O(logn)
    Space Complexity: O(1)
    """
    l,r=0, len(array)-1

    if l==r and array[l] == num:
        return True 

    while l<=r:
        print(l,r)
        mid = (l+r) // 2
        if array[mid] > num:
            r = mid - 1
        elif array[mid] < num:
            l = mid + 1 
        elif array[mid] == num:
            return mid
    return False
```

```python
binary_search(arr, n)
```

# Linked List

```python
class node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class LinkedList():
    def __init__(self):
        self.head = None
        self.size: int = 0
    def append(self, new_node: node):
        if self.size == 0:
            self.head = new_node
        else:
            curr = self.head
            while curr.next:
                curr = curr.next
            curr.next = new_node
        self.size += 1
    def __str__(self):
        if self.size == 0:
            return "Zero Elements in Linked List."
        else:
            print_ll = ""
            curr = self.head
            while curr.next:
                print_ll += str(curr.data) + " -> "
                curr = curr.next
            print_ll += str(curr.data) + " -> NONE"
            return print_ll
```
