```python
arr = [23] #, 66, 82, 99, 100, 121, 222]
n = 23
```

```python
def binary_search(array, num):
    """
    Time Complexity: O(logn)
    Space Complexity: O(1)
    """
    l,r=0, len(array)-1

    if l==r and array[l] == num:
        return True 

    while l<=r:
        print(l,r)
        mid = (l+r) // 2
        if array[mid] > num:
            r = mid - 1
        elif array[mid] < num:
            l = mid + 1 
        elif array[mid] == num:
            return mid
    return False
```

```python
binary_search(arr, n)
```
